- observar a distribuição de valores na coluna alvo (Class) e plotar essa distribuição (com a numeração da quantidade de registros + porcentagem, usando eng_fmt)


```python
df["Class"].value_counts(normalize=True)
```

```python
fig, ax = plt.subplots()

eng_fmt = EngFormatter(places=0)

sns.countplot(data=df, x="Class", ax=ax)

ax.bar_label(ax.containers[0], label_type="edge", weight="bold", labels=map(eng_fmt.format_eng, df["Class"].value_counts()))
ax.bar_label(ax.containers[0], label_type="center", weight="bold", labels=map(lambda num: f"{num:.2%}", df["Class"].value_counts(normalize=True)),
             color="white", fontsize=10)

ax.set_title("Distribuição das classes")
ax.set_xlabel("Classe")
ax.set_ylabel("Quantidade")
ax.set_yscale("log")
ax.grid(False)
#plt.gca().get_yaxis().set_visible(False)
plt.show()
```
