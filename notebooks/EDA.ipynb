#### Quais variáveis apresentam uma diferença estatisticamente significativa entre as duas classes?

```python
features = df.columns.drop("Class").to_list()

coluna_target = "Class"

analise_mw = {}

classes = df[coluna_target].unique()

alfa = 0.01

agrupamento = df.groupby(coluna_target)
```

```python
for coluna in features:
    grupo0 = agrupamento.get_group(classes[0])[coluna]
    grupo1 = agrupamento.get_group(classes[1])[coluna]

    estatistica_mw, pvalue_mw = mannwhitneyu(grupo0, grupo1)

    analise_mw[coluna] = {"pvalue":pvalue_mw, "estatistica":estatistica_mw, "relacao_significativa":pvalue_mw < alfa}

df_mw = pd.DataFrame(analise_mw).T
df_mw
```

```python
df_mw.query("relacao_significativa == False")
```

```python
df_majoritario = df[df["Class"] == 0]
df_minoritario = df[df["Class"] == 1]
```

```python
df_majoritario.shape
```

```python
df_minoritario.shape
```

```python
def remover_outliers_quantil(df, variaveis, quantil_inferior=0.05, quantil_superior=0.95):
    df_filtrado = df.copy()
    
    for variavel in variaveis:
        limite_inferior = df[variavel].quantile(quantil_inferior)
        limite_superior = df[variavel].quantile(quantil_superior)
        df_filtrado = df[(df[variavel] >= limite_inferior) & (df[variavel] <= limite_superior)]
    
    return df_filtrado
```

```python
variaveis = df.columns.drop("Class")
df_majoritario_filtrado = remover_outliers_quantil(df_majoritario, variaveis)

df_sem_outliers = pd.concat([df_majoritario_filtrado, df_minoritario])
df_sem_outliers = df_sem_outliers.reset_index(drop=True)
```
